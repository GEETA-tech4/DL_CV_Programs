```python
p = 23
g = 5

a = 6
b = 15

A = (g ** a) % p
B = (g ** b) % p

secret_key_A = (B ** a) % p
secret_key_B = (A ** b) % p

print("Public Key of A :", A)
print("Public Key of B :", B)

print("Secret Key for A :", secret_key_A)
print("Secret Key for B :", secret_key_B)
```
