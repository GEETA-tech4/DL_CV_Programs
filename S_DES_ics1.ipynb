```python
### S-DES Program using XOR ###

plaintext = "10101010"
key = "11001100"

ciphertext = ""

for i in range(len(plaintext)):

    if plaintext[i] == key[i]:
        ciphertext += "0"
    else:
        ciphertext += "1"

print("Plaintext  :", plaintext)
print("Key        :", key)
print("Ciphertext :", ciphertext)
```
