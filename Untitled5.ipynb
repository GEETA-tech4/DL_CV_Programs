```python
# Short Simple S-AES Encryption Program

plaintext = "1101011100101000"
key = "0100101011110101"

# Substitute bits
sub = ""
for b in plaintext:
    if b == "0":
        sub += "1"
    else:
        sub += "0"

# Encryption
cipher = ""
for i in range(len(sub)):
    if sub[i] == key[i]:
        cipher += "0"
    else:
        cipher += "1"

print("Plaintext :", plaintext)
print("Key       :", key)
print("Ciphertext:", cipher)
```
