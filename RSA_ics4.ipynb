```python
# Simple RSA Algorithm

# Step 1: Choose two prime numbers
p = 3
q = 11

# Step 2: Calculate n
n = p * q

# Step 3: Calculate phi
phi = (p - 1) * (q - 1)

# Step 4: Choose e
e = 7

# Step 5: Choose d
d = 3

# Message
msg = 5

# Encryption
cipher = (msg ** e) % n

# Decryption
plain = (cipher ** d) % n

# Output
print("Original Message :", msg)
print("Encrypted Message :", cipher)
print("Decrypted Message :", plain)
```
